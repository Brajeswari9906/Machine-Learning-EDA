2. Explain overfitting and underfitting in ML. How can you detect and prevent them? (Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques.)

    - In Machine Learning (ML), a model should learn patterns from training data and make accurate predictions on new data. However, sometimes the model may learn too little or too much from the data, leading to underfitting or overfitting. These problems affect the performance and accuracy of the model.

    - Underfitting occurs when a machine learning model is too simple to learn the underlying patterns in the training data. As a result, the model performs poorly on both training and testing data.Characteristics of underfitting are High training error,High testing error,Model fails to capture important relationships. Causes are Very simple model, Insufficient training time,Lack of relevant features,Excessive regularization.Example:Using a linear model to represent highly complex nonlinear data may lead to underfitting.Prevention methods are Use a more complex model, Add more useful features,Reduce regularization,Increase training time,Use feature engineering techniques.

    - Overfitting occurs when a model learns the training data too well, including noise and unnecessary details. The model performs very well on training data but poorly on unseen testing data.Characteristics are Very low training error, High testing error,Poor generalization ability.Causes are Highly complex model,Small training dataset,Too many parameters,Training for too long.Example:A deep neural network memorizing training examples instead of learning general patterns.Prevention methods are Use more training data, Apply regularization techniques,Reduce model complexity,Use dropout in neural networks,Perform early stopping,Use cross-validation.

      The bias-variance tradeoff explains the balance between underfitting and overfitting.
      
      Bias refers to errors caused by overly simple assumptions in the learning algorithm.

        - High bias leads to underfitting.
        - Model cannot capture data complexity.

      Variance refers to how much the model changes with different training data.

        - High variance leads to overfitting.
        - Model becomes too sensitive to training data noise.

      An ideal model should have:

        - Low bias
        - Low variance
        - Good generalization performance

      The goal is to find the right balance between model simplicity and complexity.

      Cross-validation divides data into multiple parts and tests the model repeatedly.

        - Helps measure model generalization.
        - Detects whether the model is stable or overfitting.

    - Regularization reduces model complexity and prevents overfitting.

    1) L1 Regularization (Lasso)
      - Adds penalty based on absolute values of coefficients.
      - Can remove unnecessary features.

    2) L2 Regularization (Ridge)
      - Adds penalty based on squared coefficients.
      - Reduces large parameter values.

    3) Dropout
      - Randomly removes neurons during training in deep learning models.
      - Prevents dependency on specific neurons.

    4) Early Stopping
      - Stops training when validation performance starts decreasing.
      - Prevents excessive learning of noise.

3. How would you handle missing values in a dataset? Explain at least three
methods with examples.
Hint: Consider deletion, mean/median imputation, and predictive modeling.

    - Missing values are common in datasets and can affect the accuracy and performance of machine learning models.Missing data may occur due to human errors, sensor failures, incomplete surveys, or data corruption. Therefore, handling missing values properly is an important step in data preprocessing. Deletion method-In this method, rows or columns containing missing values are removed from the dataset.
        - Row Deletion: Remove records with missing values.
        - Column Deletion: Remove features with too many missing values.

    - Advantages
        - Simple and easy to apply
        - Useful when missing values are very few

    - Disadvantages
        - Loss of important information
        - Not suitable for large amounts of missing data

    - Mean or Median Imputation-Missing values are replaced using statistical measures such as mean or median.

    - Mean Imputation: Replace missing values with the average of available values.

    - Median Imputation: Replace missing values with the median value.

    - Median is preferred when the dataset contains outliers.

    - Mean Imputation
      - Mean = (20 + 22 + 24) / 3 = 22

        The missing value is replaced with 22.

    - Median Imputation

      - Median of 20, 22, 24 = 22

        The missing value is also replaced with 22.

    - Advantages
      - Easy and fast method
      - Preserves dataset size

    - Disadvantages
      - Can reduce data variability
      - May introduce bias

    Predictive Modeling Imputation-In this method, machine learning algorithms are used to predict missing values based on other features in the dataset.

    Common models used:

    - Linear Regression
    - Decision Trees
    - K-Nearest Neighbors (KNN)

    - Advantages
      - More accurate than simple imputation
      - Maintains relationships between variables

    - Disadvantages
      - Computationally expensive
      - Requires proper model selection

4. What is an imbalanced dataset? Describe two techniques to handle it(theoretical + practical). Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models.
    - An imbalanced dataset is a dataset in which one class contains significantly more samples than another class. In such cases, machine learning models may become biased toward the majority class and perform poorly on the minority class.

    - Imbalanced datasets are common in real-world applications such as fraud detection, medical diagnosis, spam detection, and rare event prediction.

    - Problems Caused by Imbalanced Data
      - Model becomes biased toward the majority class
      - Poor prediction for minority class
      - High accuracy but low real performance
      - Important rare cases may be ignored

    For example, a model predicting all transactions as “non-fraud” would still achieve 95% accuracy, but it completely fails to detect fraud cases.

    - Technique 1 - Random Oversampling
      
      Theoretical Explanation - Random Oversampling increases the number of minority class samples by duplicating existing minority records. This helps balance the dataset and gives equal importance to both classes during training.

      Advantages
      - Simple to implement
      - Prevents loss of information
      - Improves minority class learning

      Disadvantages
      - Can cause overfitting because duplicated data is repeatedly learned
      - Technique 2: SMOTE (Synthetic Minority Oversampling Technique)


      Theoretical Explanation - SMOTE is an advanced oversampling technique that creates synthetic minority class samples instead of simply duplicating existing ones.

      It works by:

      - Selecting minority class samples
      - Finding nearest neighbors
      - Generating new artificial data points between them

      This increases diversity in the minority class and reduces overfitting.

      Advantages
      - Reduces overfitting
      - Creates more diverse training data
      - Improves model generalization

      Disadvantages
      - Can generate noisy synthetic samples
      - Computationally more expensive than random oversampling

5. Why is feature scaling important in ML? Compare Min-Max scaling and Standardization. Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.
    - Feature scaling is a data preprocessing technique used to bring different features of a dataset into a similar range or scale. In many machine learning algorithms, features may have different units and magnitudes. If the data is not scaled properly, features with larger values may dominate the learning process and reduce model performance.
    
    For example:
    - Age ranges from 20 to 60
    -Salary ranges from 20,000 to 1,00,000

    Here, salary values are much larger than age values, which may bias the model.
    Featuring scaling is important as,
    (1) Improves Distance-Based Algorithms

    Algorithms such as K-Nearest Neighbors (KNN), Support Vector Machine (SVM), and K-Means clustering depend on distance calculations.

    If one feature has very large values, it dominates the distance calculation and affects prediction accuracy.

      Example:-

      Suppose two features are:

      Height = 170 cm
      Salary = 80,000

      Salary will dominate the Euclidean distance unless scaling is applied.

      (2) Faster Gradient Descent Convergence

      Algorithms using gradient descent, such as Linear Regression, Logistic Regression, and Neural Networks, converge faster when features are scaled.

      Without scaling:

      - The cost function becomes uneven
      - Gradient descent moves slowly
      - Training takes more time

      With scaling:

      - Optimization becomes smoother
      - Faster convergence occurs

6. Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other? Hint: Consider categorical variables with ordinal vs. nominal relationships.
    - Machine learning algorithms cannot directly process categorical data such as names, colors, or categories. Therefore, categorical variables must be converted into numerical form using encoding techniques. Two commonly used encoding methods are Label Encoding and One-Hot Encoding.

    (1) Lable encoding-Label Encoding converts each category into a unique numerical value.

      Characteristics
      - Each category gets a unique integer.
      - Easy and memory-efficient.
      - Creates an ordinal relationship between categories.

      Advantages
      - Simple and fast
      - Requires less memory
      - Suitable for ordinal data

      Disadvantages
      - The model may assume numerical order even when none exists.
      - Can mislead distance-based algorithms.

    (2) One-Hot Encoding converts each category into separate binary columns.

      Characteristics
      - No ordinal relationship is created.
      - Categories are treated equally.
      - Increases the number of columns.

      Advantages
      - Prevents false ordering
      - Works well for nominal data
      - Improves performance of many ML models

      Disadvantages
      - Creates many columns for large datasets
      - Increases memory usage

      When to Prefer Each Method-

      Prefer Label Encoding When:
      - Categories have meaningful order.
      - Dataset has many categories and memory efficiency is important.
      - Tree-based algorithms are used, such as Decision Trees or Random Forests.

      Prefer One-Hot Encoding When:
      - Categories are nominal with no order.
      - Using distance-based algorithms such as KNN or SVM.
      - Avoiding false numerical relationships is important.


7. Google Play Store Dataset

    a). Analyze the relationship between app categories and ratings. Which categories have the highest/lowest average ratings, and what could be the possible reasons?

    Dataset: https://github.com/MasteriNeuron/datasets.git
    (Include your Python code and output in the code box below.)


```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("googleplaystore.csv")

# Check first rows
print(df.head())

# Remove missing ratings
df = df.dropna(subset=['Rating'])

# Group by Category and calculate average rating
avg_rating = df.groupby('Category')['Rating'].mean()

# Sort values
avg_rating = avg_rating.sort_values(ascending=False)

# Display results
print("\nAverage Ratings by Category:\n")
print(avg_rating)

# Highest rated category
highest = avg_rating.idxmax()
highest_value = avg_rating.max()

# Lowest rated category
lowest = avg_rating.idxmin()
lowest_value = avg_rating.min()

print("\nHighest Rated Category:")
print(highest, ":", round(highest_value, 2))

print("\nLowest Rated Category:")
print(lowest, ":", round(lowest_value, 2))
```

**Analysis of App Categories and Ratings**

- The analysis was performed on the Google Play Store dataset by grouping apps according to their categories and calculating the average rating for each category.

The results show that app ratings vary across different categories.

**Highest Rated Categories**

1. EVENTS
2. EDUCATION
3. ART_AND_DESIGN
4. BOOKS_AND_REFERENCE

Possible Reasons

- These apps usually provide focused and useful services.
Users find them easy to use and informative.
They often contain fewer ads and less intrusive features.
Educational and creative apps tend to satisfy user expectations more consistently.

**Lowest Rated Categories**

1. DATING
2. TOOLS
3. SOCIAL

Possible Reasons

- Users may experience excessive advertisements.
Privacy and security concerns are common in dating and social apps.
Tool apps often face complaints related to bugs, battery usage, or poor performance.
High competition increases user expectations, leading to stricter ratings.

b) Age and survival Analysis

```python
import pandas as pd

df = pd.read_csv('/content/titanic.csv')

df_age = df.dropna(subset=['Age']).copy()

df_age['Age_Group'] = df_age['Age'].apply(
    lambda x: 'Child' if x < 18 else 'Adult'
)

age_survival = df_age.groupby('Age_Group')['Survived'].mean() * 100

print("Survival Rate by Age Group:\n")
print("Adult :", round(age_survival['Adult'], 2))
print("Child :", round(age_survival['Child'], 2))

print("\nName: Survived, dtype: float64")
```

**Analysis for Effect of Age on Survival**

The analysis indicates that children had a higher survival rate than adults.

Reasons:

*   During evacuation, the “Women and Children First” policy was followed.
*   Children received priority access to lifeboats.
*   Adults, especially men, were more likely to stay behind during rescue operations.

10. HR Analytics Dataset

    a) What factors most strongly correlate with employee attrition? Use visualizations to show key drivers (e.g., satisfaction, overtime, salary).

    b) Are employees with more projects more likely to leave?

    Dataset: hr_analytics

a) Employee attrition

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv('/content/hr_analytics.csv')

attrition_factors = df[['satisfaction_level',
                        'last_evaluation',
                        'number_project',
                        'average_montly_hours',
                        'time_spend_company',
                        'left']].corr()['left']

print("Correlation with Employee Attrition:\n")

for factor, value in attrition_factors.items():
    print(factor, ":", round(value, 2))

print("\nName: left, dtype: float64")
```

**Analysis for factors affecting employee attrition **

This analysis shows that employee attrition is strongly related to:

1. Satisfaction level
2. Number of projects
3. Average monthly hours
4. Time spent in company

Employees with low satisfaction levels, excessive working hours, and high project loads are more likely to leave the company.

b) Employees with more projects

```python
import matplotlib.pyplot as plt

project_attrition = df.groupby('number_project')['left'].mean() * 100

print("Employee Attrition by Number of Projects:\n")

for project, rate in project_attrition.items():
    print("Projects", project, ":", round(rate, 2))

print("\nName: left, dtype: float64")

project_attrition.plot(kind='bar')

plt.title('Attrition Rate by Number of Projects')
plt.xlabel('Number of Projects')
plt.ylabel('Attrition Rate (%)')

plt.show()
```

**Analysis for employees with more projects**

The analysis indicates that employees handling a very high number of projects tend to show higher attrition rates.

Reasons:

1. Increased workload causes stress and burnout.
2. Long working hours reduce work-life balance.
3. Employees may feel overburdened without sufficient rewards or promotions.
